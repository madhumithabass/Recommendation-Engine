## Building UBCF using Euclidean distance

```python
from sklearn.metrics import pairwise_distances
```

```python
euclidean_piv_table= pd.pivot_table(data =movie_data ,values="rating",  index="userId", columns="movie",fill_value=0)
euclidean_piv_table.columns = movie_data["movie"].unique()
euclidean_piv_table.index = movie_data["userId"].unique()
euclidean_piv_table
```

```python
euclidean_correlation = pd.DataFrame(pairwise_distances(X = euclidean_piv_table,metric='euclidean'))
euclidean_correlation.columns = movie_data["userId"].unique()
euclidean_correlation.index = movie_data["userId"].unique()
euclidean_correlation
```

## Building UBCF using Cosine Similarity

```python
cosine_correlation = pd.DataFrame(1 - pairwise_distances(X = euclidean_piv_table,metric='cosine'))
cosine_correlation.columns = movie_data["userId"].unique()
cosine_correlation.index = movie_data["userId"].unique()
cosine_correlation
```
